## **Pandas_Matplotlib Notebook**

### **Definition**:

Matplotlib turns numbers into pictures — line charts, bar charts, histograms, scatter plots, pie charts.

**Why learn it?**
- It is the base library that most other Python chart tools are built on
- Works in notebooks, scripts, and websites
- Full control over how a chart looks

**When to use it vs other tools:**
| Situation | Best Tool |
|---|---|
| Quick chart, full control | Matplotlib |
| Statistical chart, less code | Seaborn |
| Chart straight from a table of data | Pandas **.plot()** |
| Interactive chart (hover, zoom) | Plotly / Bokeh |

# Matplotlib Line Styles and Markers

In Matplotlib, we can change the **line style** and **marker style** using `linestyle` and `marker`.

## 1. Line Styles

| Linestyle | Meaning |
|---|---|
| **'-'** | Solid line |
| **'--'** | Dashed line |
| **'-.'** | Dash-dot line |
| **':'** | Dotted line |
| **'None'** | No line |

## 2. Marker Styles

| Marker | Meaning |
|---|---|
| **'o'** | Circle |
| **'s'** | Square |
| **'^'** | Triangle up |
| **'v'** | Triangle down |
| **'*'** | Star |
| **'+'** | Plus |
| **'x'** | X |
| **'D'** | Diamond |
| **'.'** | Point |



```python
# Coding
week = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [200, 350, 400, 200, 500]

plt.plot(week, sales,
         color= "blue",
         linestyle = ":",
         marker = '+',
         markersize = 10)        # If you change the markersize, you change the size of mark dot.
plt.title("Sales of the Week")
plt.xlabel("Days of Week")
plt.ylabel("Sales of the week")
plt.show()
```

**What changed:** same line, but now it is green, dashed, and shows a dot at every data point.

### **Two Lines on the Same Chart**

Just call **plt.plot()** a second time before **plt.show()**.

```python
# Coding
product_a = [10,20,30,20,15,10,5]
product_b = [5,10,20,25,30,20,30]

plt.plot(product_a,
         marker = 'o',
         label= 'Product A')
plt.plot(product_b,         # It shows second line
         marker = "s",
         label= 'Product B')
plt.legend()        # Is shows which color is which product
plt.title("Product Sales Comparision")
plt.xlabel("Day")
plt.ylabel("Units Sold")
plt.show()
```

**What changed:** a second **plt.plot()** line was added, plus **plt.legend()** so we know which line is which.

### **Bar Chart**

Same idea as a line chart, but use **plt.bar()** instead of **plt.plot()**.

```python
# Coding
categories = ['A', 'B', 'C', 'D', 'E', 'F', 'G']
values = [10, 20, 20, 30, 40, 30, 60]

plt.bar(categories, values,
        color = ['red', 'blue', 'green', 'yellow', 'orange', 'purple', 'pink']
        )
plt.title("Comparision of Categories")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
# Coding
categories = ['A', 'B', 'C', 'D', 'E', 'F', 'G']
values = [10, 20, 20, 30, 40, 30, 60]

plt.bar(categories, values,
        color = ['red', 'blue', 'green', 'yellow', 'orange', 'purple', 'pink'],
        edgecolor = "black",
        linewidth = 1,      # Linewidth change the size of line around the bar; thicker to thinner.
        width = 0.5,        # width change the size of bar; wider to narrow.
        alpha = 0.5)        # alpha change the brightness of chart from lighter to darker.
plt.title("Comparision of Categories")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

## **Pie Chart**

Use **plt.pie()** to show parts of a whole (use only 2-6 slices).

```python
# Coding
students = ["Mann", 'James', "lilly", 'Patrick', 'Tony', 'Peter', "Mike"]
marks = [80, 70, 90, 65, 68, 88, 60]

plt.pie(marks, labels=students, autopct = "%1.1f%%")
plt.title("Marks Distribution")
plt.show()
```

Actually, the important part is:

% → start of formatting
1.1 → display 1 decimal place
f → floating-point number
%% → print the actual % symbol

## **Histogram**

A histogram shows how data is **spread out** (how many values fall in each range).



### Create the data

First, create some sample data.

```python
# Coding
import matplotlib.pyplot as plt

marks = [35, 38, 42, 43, 44, 46, 48, 49, 50,
         51, 53, 54, 56, 58, 60, 62, 64]

bins = [30,35, 40,45, 50,55, 60,65, 70]

plt.hist(marks, bins = bins,
         edgecolor = "blue",  # edgecolor is boder color
         color = "skyblue",
         alpha = 0.5)       # alpha change the brightness of chart from lighter to darker.

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.grid(axis = 'y',         # The grid lines will appear behind the background and it can be horizontal that is axis = y
         )
plt.show()

```

## **Scatter Plot**

A scatter plot shows the relationship between **two** numeric variables — one dot per data pair.



### Create the data

First, create some sample data. Create two lists of the same length — hours studied and the exam score for 8 students.

```python
# Coding
hours_studied = [1,2,3,4,5,6,7,8]
exam_score = [50,55,65,60,70,75,80,90]

print(list(zip(hours_studied, exam_score)))      # peek at the pairs
```

**What we have:** 8 pairs of numbers — student 1 studied 1 hour and scored 50, student 2 studied 2 hours and scored 55, and so on. Nothing plotted yet.

### The most basic scatter plot
call **plt.scatter(x, y)**

```python
# Coding
plt.scatter(hours_studied, exam_score)
plt.show()
```

### Adding a Color in Scatter plot
Add color to change the dot from color- blue to other random color.

```python
# Coding
plt.scatter(hours_studied, exam_score,
            color = "purple")         # NEW: purple dots
plt.show()
```

### Changing dot size

Add **s** (size) to make the dots bigger or smaller.

```python
# Coding
plt.scatter(hours_studied, exam_score,
             color = "orange",
             s= 50)         # NEW: s=100 makes dots bigger
plt.show()
```

### Add title and axis labels

Same as always — make it readable.

```python
# Coding
plt.scatter(hours_studied, exam_score,
            color = "Orange",
            s = 50)
plt.title("Study Hours VS Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()
```

**What changed:** now anyone looking at the chart knows exactly what it's measuring.

### Color dots by a third value

Instead of one flat color, pass a **list of numbers** to **c=** — matplotlib will color each dot based on its value, like a heat scale.

```python
# Coding
plt.scatter(hours_studied, exam_score,
            c = exam_score,         # NEW: color each dot by its exam score
            s = 100,
            cmap = "viridis")         # NEW: cmap = the color scale to use
plt.title("Study Hours vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.colorbar(label= "exam_score")     # NEW: shows what the colors mean
plt.show()
```

**What changed:** instead of every dot being the same purple, low scores are now a different color than high scores, and a colorbar explains the scale.

### Compare two groups on one scatter plot

Just like the histogram, add a second **plt.scatter()** call to compare two groups.

```python
# Coding
group_a_hours = [1,2,3,4,5]
group_a_scores = [50,55,60,68,75]

group_b_hours = [4,5,6,7,8]
group_b_scores = [70,75,80,85,90]

group_c_hours = [7,8,9,10,11]
group_c_scores = [60,50,55,70,65]

plt.scatter(group_a_hours, group_a_scores,
            color = "orange",
            s = 100,
            label = "Group A")
plt.scatter(group_b_hours, group_a_scores,
            color = "blue",
            s = 100,
            label = "Group B")
plt.scatter(group_c_hours, group_c_scores,
            color = "purple",
            s = 100,
            label = "Group C")

plt.title("Study Hours VS Exam Score: Group A VS Group B VS Group C")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.legend()
plt.show()
```

**What changed:** a second **plt.scatter()** call was layered on the same chart, with its own color and label, so we can compare the two classes directly.

### **Saving a Chart to a File**

Call **plt.savefig()** **before** **plt.show()**.

```python
# Coding
plt.scatter(hours_studied, exam_score,
            color = "purple",
            s = 100)
plt.title("Study Hours VS Exam Score")
plt.savefig("Scatter_chart.png", dpi= 200, bbox_inches = "tight")
plt.show()
print("Saved as scatter_chart.png")
```

### **Alternative Libraries — and When to Use Them**

| Library | Best For | Value |
|---|---|---|
| **Matplotlib** | Full control, static charts | Most flexible, works everywhere |
| **Seaborn** | Statistical plots (heatmaps, box plots) | Prettier defaults, less code |
| **Pandas **.plot()**** | Quick chart from a table of data | Fastest for data you already loaded |
| **Plotly** | Interactive charts, dashboards | Hover, zoom, web-ready |
| **Bokeh** | Big data, interactive dashboards | Handles large datasets well |
| **Altair** | Short, declarative chart code | Very readable syntax |

**Simple rule:** exploring data → Pandas. Statistics → Seaborn. Need interactivity → Plotly/Bokeh. Need full control → Matplotlib.
