### **Create the 2x2 grid subplot of **Line Graph**, **Pie Chart**, **Double Line Graph** , **Scatter Plot** and your color customization with xlabel, ylabel, title.**

````python
# Coding
# For Line Graph:
employee = ["James", "Tony", "Lilly", "Mann", "Patt", "Peter", "Jude"]
salary = [2000, 3000, 1500, 3000, 4000, 3000, 5000]

plt.subplot(2,2,1)
plt.plot(employee, salary,
         color = "skyblue",
         marker = "o",
         linewidth = 1)
plt.title("Line Graph")
plt.xlabel = "Employee"
plt.ylabel = "Salary"

# For Pie Chart:
expenses = ['Food', 'Clothes', 'Rent', 'Study', 'Travel']
amount = [1000, 1500, 4000, 5000, 1000]

plt.subplot(2,2,2)
plt.pie(amount, labels=expenses, autopct= "%1.1f%%")
plt.title("Pie Chart")

# For Double Line Graph:
product_a = [10, 20, 15, 30, 25]
product_b = [5, 15, 25, 10, 20]

plt.subplot(2,2,3)
plt.plot(product_a, color = "skyblue", label = "Product A")
plt.plot(product_b, color = "orange", label = "Product B")
plt.title("Double Line Graph")
plt.legend()


# For Scatter Plot:
hours_studied = [1, 2, 3, 4, 5, 6, 7, 8]
exam_score =    [50, 55, 65, 60, 70, 75, 85, 90]

plt.subplot(2,2,4)
plt.scatter(hours_studied, exam_score, color = "green")
plt.title("Scatter Plot")

plt.tight_layout()
````

### **Indexing in subplot to access specific plots**:

We access them using indexing for individual plots:

axes[0, 0] → top-left

axes[0, 1] → top-right

axes[1, 0] → bottom-left

axes[1, 1] → bottom-right

### **Create a 2x2 subplot grid by given data:**

````python
# Coding
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

fig, axes = plt.subplots(2,2)

# Top - Left Position
axes[0,0].plot(x,y)
axes[0,0].set_title("Plot 1")

# Top - Right Position
axes[0,1].plot(x,y)
axes[0,1].set_title("plot 2")

# Bottom - Left Position
axes[1,0].plot(x,y)
axes[1,0].set_title("Plot 3")

# Bottom - Right Position
axes[1,1].plot(x,y)
axes[1,1].set_title("Plot 4")

plt.tight_layout()
````

## Difference Between **plt.subplot()** and **plt.subplots()**

| **plt.subplot()** | **plt.subplots()** |
|---|---|
| Selects one subplot at a time | Creates the Figure and Axes together |
| Uses **plt.subplot(rows, columns, position)** | Uses **fig, axes = plt.subplots(rows, columns)** |
| We usually use **plt.plot()** | We use **axes.plot()** |
| Works with the **current Axes** | Gives us direct access to each **Axes object** |
| Example: **plt.subplot(2, 2, 1)** | Example: **fig, axes = plt.subplots(2, 2)** |
| Good for learning the basic subplot concept | Better for managing multiple plots |
| Convenient for simple plots | More flexible for complex plots |

### **Example**

#### Using **plt.subplot()**

```python
plt.subplot(2, 2, 1)
plt.plot(x, y)
```

#### Using **plt.subplots()**

```python
fig, axes = plt.subplots(2, 2)
axes[0, 0].plot(x, y)
```

Both draw the same line in the same top-left position — **plt.subplots()** just hands you
the **axes** object directly instead of making you re-select it with **plt.subplot()** each time.


### **Controlling Figure Size with figsize**

So far every figure has used Matplotlib's default size. **figsize=(width, height)** (in inches)
lets us make the whole figure bigger or smaller — useful when subplots start looking cramped.

### **Create a 2x2 grid, but make the whole figure larger**

````python
# Coding

import matplotlib.pyplot as plt
fig, axes= plt.subplots(2,2, figsize= (10,8))   # This figsize will expand the size of graph as per the input number.

# Plot 1
axes[0,0].plot([1,2,3], [2,4,6])
axes[0,0].set_title("Plot 1")

# Plot 2
axes[0,1].plot([1,2,3], [10,20,30])
axes[0,1].set_title("Plot 2")

# Plot 3
axes[1,0].plot([1,2,3], [100,200,300])
axes[1,0].set_title("Plot 3")

# Plot 4
axes[1,1].plot([1,2,3], [1000,2000,3000])
axes[1,1].set_title("Plot 4")

plt.tight_layout()
````

## **Manual Spacing with subplots_adjust**

**plt.tight_layout()** adjusts spacing automatically, but sometimes we want exact control.
**plt.subplots_adjust()** sets the spacing ourselves using **hspace** (height) and **wspace** (width).

````python
# Coding
fig, axes = plt.subplots(2,2, figsize = (8,6))

axes[0,0].plot([1,2,3], [2,4,6]); axes[0,0].set_title("Plot 1")
axes[0,1].plot([1,2,3], [20,40,60]); axes[0,1].set_title("Plot 2")
axes[1,0].plot([1,2,3], [200,400,600]); axes[1,0].set_title("Plot 3")
axes[1,1].plot([1,2,3], [100,200,300]); axes[1,1].set_title("Plot 4")

plt.subplots_adjust(hspace = 0.5, wspace= 0.4)   # This gives the space between each plots manually as per the input hspace and wspace.
````

## **Uneven Layouts with GridSpec**

**plt.subplots()** always gives an evenly sized grid. When we need one plot to be wider or
taller than the others — like one big plot on top and two small ones below — **GridSpec**
lets us design the layout ourselves.

````python
# Coding
import matplotlib.gridspec as gridspec

fig = plt.figure(figsize= (8,6))
gs = gridspec.GridSpec(2,2, figure = fig)

# Top row: one wide plot spanning both columns
ax1 = fig.add_subplot(gs[0,:])
ax1.plot([1,2,3,4], [10, 20, 15, 25])
ax1.set_title("Wide Plot (Span both columns)")

# Bottom row: two  smaller plots, one per column
ax2 = fig.add_subplot(gs[1,0])
ax2.bar(["A", "B", "C"], [10, 20 , 15])
ax2.set_title("Bottom Left")

ax3 = fig.add_subplot(gs[1,1])
ax3.scatter([1,2,3], [3,1,2])
ax3.set_title("Bottom Rght")

plt.tight_layout()
plt.show()
````

## **Saving a Multi-Subplot Figure**

**fig.savefig()** saves the WHOLE figure — every subplot in it — to a file. **dpi** controls
the resolution, and **bbox_inches="tight"** trims extra whitespace around the edges.

````python
# Coding
fig, axes = plt.subplots(1,2, figsize = (8,4))

axes[0].plot([1,2,3], [10,20,30])
axes[0].set_title("Students Marks")

axes[1].bar(["Jan", "Feb", "Mar"], [100, 200, 150])
axes[1].set_title("Sales of the Month")

plt.tight_layout()
fig.savefig("dashboard.png", dpi= 150, bbox_inches = "tight")
plt.show()

print("Saved dashboard.png")
````

## **Quick Reference: Subplot Tools**

| Function | What it does |
|---|---|
| **plt.subplot(rows, cols, pos)** | Selects one plot at a time in a grid |
| **plt.subplots(rows, cols)** | Creates the figure and all Axes together |
| **figsize=(w, h)** | Sets the overall figure size in inches |
| **sharex=True** / **sharey=True** | Shares an axis scale across subplots |
| **fig.suptitle("...")** | Adds one title above the whole figure |
| **plt.tight_layout()** | Adjusts spacing automatically |
| **plt.subplots_adjust(hspace, wspace)** | Adjusts spacing manually |
| **GridSpec** | Builds grids with differently sized plots |
| **fig.savefig("name.png")** | Saves the entire figure, all subplots included |
